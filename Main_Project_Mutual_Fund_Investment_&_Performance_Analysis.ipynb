### **4.Feature Engineering** (creating new column)


**1.Flow_Status** (new column 1)

```python
#creating a new useful column:
#using net_inflow_outflow column and find net_flow_status

def flow_status(x):
  if x > 0:
    return "Positive"
  elif x < 0:
    return "Negative"
  else:
    return "No Flow"

df["net_flow_status"] = df["net_inflow_outflow"].apply(flow_status)
```

**2.Investment Status** (new column 2)

comparing fund mobilization and redemptions. So we can identify whether mobilization is greater than redemption.

```python
#Compare fund mobilization and redemption to identify categories with positive and negative net inflows

def investment_status(row):
  if row["fund_mobilized"] > row["repurchase_redemption"]:
    return "More Mobilization"
  elif row["fund_mobilized"] < row["repurchase_redemption"]:
    return "More Redemption"
  else:
    return "Equal"

df["investment_status"] = df.apply(investment_status, axis = 1)
```

# **Statistical analysis**

**Measure of central Tendency + Chart**

This chart helps compare Mean, Median, and Mode visually.

Mean shows overall average, Median shows middle value.

Mode shows most frequent value (often zero in this dataset).


```python
# Measure of Central Tendency

numeric_cols = ["net_aum","avg_net_aum","net_inflow_outflow","no_of_folios","fund_mobilized","repurchase_redemption"]

summary_table = pd.DataFrame({"Mean": df[numeric_cols].mean(),"Median": df[numeric_cols].median(),
                                 "Mode": df[numeric_cols].mode().iloc[0]})

# Display the table
print("MEASURE OF CENTRAL TENDENCY\n", summary_table)
```

```python
for col in numeric_cols:
    # Directly calculate metrics for each column
    mean_val = df[col].mean()
    median_val = df[col].median()
    mode_val = df[col].mode().iloc[0]
    plt.figure(figsize=(6, 4))
    sns.histplot(df[col], kde=True, bins=30, color="skyblue")
    plt.axvline(mean_val, color="red", label=f"Mean: {mean_val:.2f}")
    plt.axvline(median_val, color="green", label=f"Median: {median_val:.2f}")
    plt.axvline(mode_val, color="orange", linestyle=":", label=f"Mode: {mode_val:.2f}")
```

**Interpretation:**

The histograms show that most of the numerical variables are highly right-skewed. The majority of observations are concentrated at lower values, while a small number of observations have very high values, creating long right tails. In these distributions, the Mean is generally higher than the Median and Mode because the extreme high values pull the Mean toward the right. net_inflow_outflow is mainly concentrated around zero, with both positive and negative values and some extreme observations. Overall, the data contains significant variation and potential outliers, so the Median may be more representative than the Mean for some of these variables.

**Measure of Dispersion: Variance & Standard Deviation**

**Variance Variance shows how widely values are spread, highlighting consistency vs volatility across metrics.**



```python
# ---------- Variance calculation ----------
variance = df[numeric_cols].var().to_frame(name="Variance")
print("MEASURE OF DISPERSION: VARIANCE \n")
print(variance)

#box plot for variance
for col in numeric_cols:
    plt.figure(figsize=(4, 2))
    df.boxplot(column=col)
    plt.title(f'Box Plot of {col} (Variance Insight)')
    plt.ylabel('Values')
    plt.show()
```

**Interpretation**

Variance analysis shows wide differences across variables.

Key Insights:

no_of_folios has the highest variance (1.52 × 10¹³), indicating very uneven investor participation.

net_aum, avg_net_aum, fund_mobilized, and repurchase_redemption show similar variances (~9.5 × 10⁹ to 1.18 × 10¹⁰), reflecting considerable spread in asset and cash‑flow values.

net_inflow_outflow has the lowest variance (8.33 × 10⁷), showing relative stability.

Observation:  
Folios vary the most, AUM and cash‑flow metrics show moderate variability, and net inflows/outflows remain comparatively stable.

***Standard Deviation***

**Standard deviation tells if numbers are spread out or close to the average**

High SD → values vary a lot.

Low SD → values stay near the average.

```python
# ---------- Standard Deviation calculation----------
std = df[numeric_cols].std().to_frame(name="Standard Deviation")
print("\nMEASURE OF DISPERSION: STANDARD DEVIATION \n")
print(std)

# 1. First, calculate mean and deviation for all columns
means = df[numeric_cols].mean()
deviation = df[numeric_cols] - means

# 2. Loop through all 6 columns to plot deviation
for col in numeric_cols:
    plt.figure(figsize=(10, 4))

    # Bar chart plotting for each column's deviation
    plt.bar(range(len(deviation)), deviation[col], color="skyblue")

    # Mean line (0 reference point)
    plt.axhline(0, color="red", linestyle="--")
    plt.title(f"Deviation from Mean - {col}")
    plt.ylabel("Deviation")
    plt.xticks([])  # remove x-axis labels
    plt.tight_layout()
    plt.show()
```

**Interpretation**

1. "no_of_folios" shows the highest standard deviation (app.3.90 million), indicating extremely wide variation in folio counts across records.
2. "net_aum" (app.97,880) and "avg_net_aum" (app.99,935) have similar spreads, both around ₹1 lakh.
3. "fund_mobilized" and "repurchase_redemption" (app.108,423) are nearly identical, showing inflows and outflows fluctuate at comparable magnitudes.
4. "net_inflow_outflow" has the smallest deviation (app.9,126), far lower than the gross figures.
5. This suggests large inflows and outflows largely offset each other, keeping the net value relatively stable.
6. Unlike variance, SD is in the same units as the original data, making it directly interpretable (e.g., "AUM deviates by app.₹98,000 on average").
7. Overall, AUM and fund-flow metrics cluster in similar variability, while folio counts are far more volatile in absolute scale.

### **Skewness**

Skewness shows whether data is balanced or tilted — positive means right-heavy, negative means left-heavy.

```python
# Skewness
skewness_table = df[numeric_cols].skew().to_frame(name='Skewness')

print("MEASURES OF DISTRIBUTION : SKEWNESS\n")
print(skewness_table)


for col in numeric_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(df[col], kde=True, bins=10, color='skyblue')

    # Mean line
    mean_val = df[col].mean()
    plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean = {mean_val:.2f}')

    # Title with Skewness
    skew_val = df[col].skew()
    plt.title(f'Distribution of {col} (Skew: {skew_val:.2f})')

    plt.xlabel(col)
    plt.ylabel('Count')
    plt.legend(loc='upper right')
    plt.show()
```

### **Kurtosis**

Kurtosis shows how extreme outliers dominate a distribution — high means heavy tails, low means balanced usage.

```python
# Comparative Analysis:
# Fund Mobilization, Redemption, Net Flow and Folios by Scheme Category
# It Covers :
# Which categories receive the highest fund mobilization?	fund_mobilized?
# Which categories have the highest redemption?	repurchase_redemption?
# Which categories have positive or negative net flows?	net_inflow_outflow?
# Which categories have the highest number of folios?
metrics = ["fund_mobilized","repurchase_redemption","net_inflow_outflow","no_of_folios"]
category_data = df.groupby("scheme_category")[metrics].mean()

# Normalize values so the four metrics can be compared
category_data_norm = (category_data - category_data.min()) / (category_data.max() - category_data.min())
category_data_norm.plot(kind="bar",figsize=(10, 6))
plt.title("Investment Activity and Investor Participation by Scheme Category")
plt.xlabel("Scheme Category")
plt.ylabel("Normalized Average Value")
plt.xticks(rotation=30, ha="right")
plt.legend(loc="upper left", bbox_to_anchor=(1.0, 1.0), title="Metric",labels=["Fund Mobilized","Repurchase/Redemption","Net Inflow/Outflow","Number of Folios"])
plt.tight_layout()
plt.show()
```

Interpretation:  
The chart compares normalized averages of fund mobilization, redemption, net inflows/outflows, and folios across scheme categories.

Key Insights (with values):

Income/Debt Schemes → Highest Fund Mobilized (~ 1.0) and Repurchase/Redemption (~0.95).
👉 Heavy liquidity cycles with large inflows and outflows.

Growth/Equity Schemes → Highest Number of Folios (~ 1.0).
👉 Broad investor participation and retail adoption.

Other Schemes → Highest Net Inflow/Outflow (~ 0.85).
👉 Strong positive fund flows, especially post‑2021.

Hybrid Schemes → Moderate across all metrics (~ 0.5–0.6).
👉 Balanced role in both investor participation and fund activity.

Solution Oriented Schemes → Lowest across all metrics (< 0.2).
👉 Minimal impact on overall investment activity.

Observation:  
Debt schemes dominate liquidity, equity schemes lead in participation, other schemes show strong inflows, hybrid schemes balance both, and solution‑oriented schemes remain marginal.
